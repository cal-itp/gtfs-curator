# GTFS Diagnostics

1. be able to filter by service_date or schedule / vp / tu / sa_name with an interactive table and find all the relevant keys
2. the keys can then be used to query warehouse with partitioning and clustering keys
3. group these by date, so that only the most recent section needs to get refreshed

```python
import gcsfs
import itables
import pandas as pd

from IPython.display import Markdown

from cookbook_vars import GCS_FILE_PATH, DATE_PERIODS_DICT
```

```python
df = pd.read_parquet(
    f"{GCS_FILE_PATH}quartet_by_feed_key_categorized.parquet", 
    filesystem = gcsfs.GCSFileSystem()
)
```

```python
itables.init_notebook_mode() 
```

```python
def display_itables_with_search(df: pd.DataFrame):
    
    return itables.show(
        df,
        layout={"top1": "searchBuilder"},
        searchBuilder={
            "preDefined": {
                "criteria": [
                    {"data": "schedule_name", "condition": "=", "value": [""]}
                ]
            }
        },
        maxBytes="5MB"
    )
```

## Jul 2026 - present

```python
most_recent_date = df.service_date_end.max().date().isoformat()
display(Markdown(f"### latest service_date: {most_recent_date}"))
```

```python
display_itables_with_search(df[df.is_2026_H2 == True])
```
